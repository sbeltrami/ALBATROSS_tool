```python
import dill
```

```python
month = 'JUL'
lead = 1
```

```python
# C3S: leggere solo il primo oggetto (mm_dic), gli altri 5 non vengono letti
with open('c3s_atl3_tpr.pkl', 'rb') as f:
    mm_dic = dill.load(f)

k_atl3 = f'{month}, lead {lead} (atl3)'
k_tp   = f'{month}, lead time {lead} (tp WA)'

mm_small = {
    k_atl3: {'atl3 index members': mm_dic[k_atl3]['atl3 index members']},
    k_tp:   {'anomaly members': mm_dic[k_tp]['anomaly members']},
}

with open('c3s_mm_jul_lead1.pkl', 'wb') as f:
    dill.dump(mm_small, f)
del mm_dic
```

```python
# ORAS5: tenere solo l'indice ATL3 di luglio, con la stessa struttura
with open('oras5_atl3.pkl', 'rb') as f:
    oras5_dic = dill.load(f)

oras5_small = {'sst': {f'{month} (atl3)': {'atl3 index': oras5_dic['sst'][f'{month} (atl3)']['atl3 index']}}}
with open('oras5_atl3_jul.pkl', 'wb') as f:
    dill.dump(oras5_small, f)
```
